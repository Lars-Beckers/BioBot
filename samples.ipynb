```python
# Imports
import numpy as np
import os
import cv2
```

```python
def save_video_frames(video_path, output_dir, frame_rate=100):
    """
    Read video and save frames to specified directory.
    
    Parameters:
    - video_path: str, path to video
    - output_dir: str, directory where frames are saved
    - frame_rate: int, amount of frames between saved frames (1 = every frame is saved)
    """

    # Create output directory if not existing
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # Open video file
    video_capture = cv2.VideoCapture(video_path)
    
    # Check if video opened successfully
    if not video_capture.isOpened():
        print(f"Error opening video file: {video_path}")
        return
    
    # Get total number of frames in video
    total_frames = int(video_capture.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in the video: {total_frames}")
    
    frame_number = 0
    
    while True:
        # Read frame from video
        ret, frame = video_capture.read()
        
        # If frame was not read correctly, exit loop
        if not ret:
            break
        
        if frame_number % frame_rate == 0:
            # Construct output filename
            frame_filename = os.path.join(output_dir, f"frame_{frame_number:04d}.jpg")
                
            # Save frame as JPEG file
            cv2.imwrite(frame_filename, frame)
            print(f"Saved frame {frame_number} to {frame_filename}")
            
        # Increment frame number
        frame_number += 1
    
    # Release video capture object
    video_capture.release()
    print(f"Frames have been saved to: {output_dir}")
```

```python
save_video_frames(video_path="data/grasrobot.mp4", output_dir="data/frames/", frame_rate=100)
```
