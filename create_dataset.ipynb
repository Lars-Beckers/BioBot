Create sample dataset of frames from video.

```python
import cv2
import os

# load variables
video_path = "dataset/grasrobot.mp4"
frames_path = "dataset/frames/"
save_frame = 100 # saves frame every X frames


# extract frames from video
video = cv2.VideoCapture(video_path) # open video
frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT)) # read amount of frames

current_frame = 0

while True:
    read, frame = video.read()
    if not read:
        break # if no frame read, end of video
    if current_frame % save_frame == 0:
        frame_name = os.path.join(frames_path, f"frame_{current_frame:04d}.jpg")
        cv2.imwrite(frame_name, frame)
        print(f"Saved frame {current_frame} to {frame_name}")
    current_frame += 1

# release video capture object
video.release()
print(f"All frames have been saved to: {frames_path}")
```
