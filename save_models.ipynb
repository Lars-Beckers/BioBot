Load pretrained models and save them as tensor.

Links:
- Plantnet300k info: https://github.com/plantnet/PlantNet-300K
- DenseNet121 model: https://pytorch.org/vision/main/models/generated/torchvision.models.densenet121.html
- EfficientNet_b0 model: https://pytorch.org/vision/main/models/generated/torchvision.models.efficientnet_b0.html
- ResNet50 model: https://pytorch.org/vision/main/models/generated/torchvision.models.resnet50.html


```python
import torch
from torchvision.models import densenet121, efficientnet_b0, resnet50


# load variables
num_classes = 1081 # number of classes in PlantNet300K
densenet_checkpoint_path = 'pretrained_models/densenet121_weights_best_acc.tar'
efficientnet_checkpoint_path = 'pretrained_models/efficientnet_b0_weights_best_acc.tar'
resnet_checkpoint_path = 'pretrained_models/resnet50_weights_best_acc.tar'


# DenseNet121
try:
    DenseNet_model = densenet121(num_classes= num_classes)
    DenseNet_decapitated = torch.nn.Sequential(*(list(DenseNet_model.children())[:-1])) # remove final layer (head)
    DenseNet_checkpoint = torch.load(densenet_checkpoint_path, map_location='cpu', weights_only=True)
    DenseNet_state_dict = DenseNet_checkpoint.get('state_dict', DenseNet_checkpoint)
    DenseNet_model.load_state_dict(DenseNet_state_dict, strict=False)
    DenseNet_decapitated.load_state_dict(DenseNet_state_dict, strict=False)
    print("DenseNet models loaded succesfully")
except Exception as e:
    print(f"Error loading weights for DenseNet: {e}")


# EfficientNet_b0
try:
    EfficientNet_model = efficientnet_b0(num_classes= num_classes)
    EfficientNet_decapitated = torch.nn.Sequential(*(list(EfficientNet_model.children())[:-1])) # remove final layer (head)
    EfficientNet_checkpoint = torch.load(efficientnet_checkpoint_path, map_location='cpu', weights_only=True)
    EfficientNet_state_dict = EfficientNet_checkpoint.get('state_dict', EfficientNet_checkpoint)
    EfficientNet_model.load_state_dict(EfficientNet_state_dict, strict=False)
    EfficientNet_decapitated.load_state_dict(EfficientNet_state_dict, strict=False)
    print("EfficientNet models loaded succesfully")
except Exception as e:
    print(f"Error loading weights for EfficientNet: {e}")


# ResNet50
try:
    ResNet_model = resnet50(num_classes= num_classes)
    ResNet_decapitated = torch.nn.Sequential(*(list(ResNet_model.children())[:-1])) # remove final layer (head)
    ResNet_checkpoint = torch.load(resnet_checkpoint_path, map_location='cpu', weights_only=True)
    ResNet_state_dict = ResNet_checkpoint.get('state_dict', ResNet_checkpoint)
    ResNet_model.load_state_dict(ResNet_state_dict, strict=False)
    ResNet_decapitated.load_state_dict(ResNet_state_dict, strict=False)
    print("ResNet models loaded succesfully")
except Exception as e:
    print(f"Error loading weights for ResNet: {e}")


# save models
# torch.save(DenseNet_model, 'saved_models/DenseNet121.pt')
torch.save(DenseNet_decapitated, 'saved_models/DenseNet121_decap.pt')
# torch.save(EfficientNet_model, 'saved_models/EfficientNet_b0.pt')
torch.save(EfficientNet_decapitated, 'saved_models/EfficientNet_b0_decap.pt')
# torch.save(ResNet_model, 'saved_models/ResNet50.pt')
torch.save(ResNet_decapitated, 'saved_models/ResNet50_decap.pt')
```
